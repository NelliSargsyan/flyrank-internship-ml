## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*


What my work will be able to say.

Observed: Through my work, I can observe, measure and say that the signals and content types move together. For example it can say that certain content types, freshness tiers, or word-count ranges are associated with better average position or CTR based on observation.

Directional: My work will be able to make comparisons between content types. It can say that a certain kind of content tends to rank better than another type. For example, it can claim that content tagged "feedly article" tends to have a higher average position than content tagged "keyword article".

Decision-support: My work can support decisions by outputting a flagged list of "signals worth investigating further" or "content worth reviewing", to help humans decide where to look next.

What my work won't be able to say.

My work can't claim that there is proof that a certain content type or a search signal causes higher or lower visibility, engagement, or clicks. Also, my work won't be able to prove a Google algorithm. It is simply going to be a pattern-recognition excercise on historical, pseudonymized data.
