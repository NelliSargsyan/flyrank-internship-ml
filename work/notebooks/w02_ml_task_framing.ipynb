## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

I changed my lane to CTR / Engagement Opportunity Scoring. I chose to do scoring implemented via a Logistic Regression (GLM, binomial. I will fit a Binomial GLM where the response is clicks_90d out of impressions_90d.  which will allow me to asign an opportunity score for each page and rank the pages by their opportunity scores. That would reveal the pages that are having good visibility in searches but not many clicks. The opportunity score comes from the model's residuals and not from it's predicted class. Logistic Regression is the right tool in this case becuase CTR is a number, a proportion between [0,1]. That is exactly what the binomial link function models.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

The metric to be used here is Precision@K. Precision@K represents the proportion of the top K pages identified by the model that are actually high opportunity. The higher it is the better. For example, if Precision@20 is 0.8, that means that out of the 20 identified pages 16 are high opportunity and that number is good.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Pages not getting many clicks but getting impressions can be influenced by many factors, such as main_intent, content_type, freshness, which a simple if statement can't take into account. A look up table (mean CTR per tier × intent × content_type) can break down on thin cells. For example, there are only a few rows that have main_intent = "navigational" and if they get further sliced by content_type and freshness there will be only 2,3 of them which is very little. A small amount of pages like that is unreliable and the lookup table doesn't show that.
